```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from collections import Counter
import pandas as pd
```

```python
fifa_df = pd.read_csv('FIFA_2019_2022.csv')
jogos = fifa_df.values
edges = [(origem, destino, {"weight": peso}) for origem, destino, peso in jogos ]
G = nx.Graph()
G_direcionado = nx.DiGraph()
G.add_edges_from(edges)
G_direcionado.add_edges_from(edges)

print(f"a) num componentes conexas = {nx.number_connected_components(G)}")

weighted_degree = dict(G_direcionado.degree(weight="weight"))
avg_weighted_degree = sum(weighted_degree.values()) / len(weighted_degree)
print(f"b) Grau ponderado médio = {avg_weighted_degree}")

maior_cc = max(nx.strongly_connected_components(G_direcionado), key=len)
subG = G.subgraph(maior_cc) 

for u, v, d in subG.edges(data=True): 
    d["dist"] = 1 / d["weight"]  # peso não é proporcional a distância
    d["dist"] = d["weight"]      # peso é proporcional a distância
 
diametro_ponderado =  nx.diameter(subG, weight="dist") 
diametro_sem_peso = nx.diameter(subG) 
print(f"c) Diametro = ponderado: {diametro_ponderado}, normal: {diametro_sem_peso}")

top10_degree = sorted(nx.out_degree_centrality(G_direcionado).items(), key=lambda x: x[1], reverse=True)[:10]
print("d) Top 10 centralidades por grau (entrando do vértice):")
for a in top10_degree:
    print(f"     {a[0]}: {a[1]:.2f}")

top10_degree = sorted(nx.pagerank(G_direcionado).items(), key=lambda x: x[1], reverse=True)[:10]
print("d) Top 10 centralidades por pageRank:")
for a in top10_degree:
    print(f"     {a[0]}: {a[1]:.2f}")
```
