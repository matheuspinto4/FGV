```python
import pandas as pd
import sqlite3
```

```python
df_sangue = pd.read_csv("bloodtypes.csv")
conn = sqlite3.connect(':memory:')
df_sangue.to_sql('blood_data', conn, index=False, if_exists='replace')

query = """
SELECT
    Country,
    CASE
        WHEN "O+" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'O+'
        WHEN "A+" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'A+'
        WHEN "B+" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'B+'
        WHEN "AB+" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'AB+'
        WHEN "O-" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'O-'
        WHEN "A-" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'A-'
        WHEN "B-" = MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") THEN 'B-'
        ELSE 'AB-'
    END AS Highest_Blood_Type,
    MAX("O+", "A+", "B+", "AB+", "O-", "A-", "B-", "AB-") AS Highest_Percentage
FROM blood_data
WHERE Highest_Percentage > 50
"""

result_df = pd.read_sql_query(query, conn)
conn.close()
result_df
```
