## **Casos de Uso**



### Diagrama de Casos de Uso

Expressa as expectativas dos stakeholders. Pode ser usado durante todo o processo de análise e clarificação de requisitos.

Responde as seguintes perguntas:

- O que está sendo descrito? 
    - Que sistema está sendo modelado?

- Quem interage com o sistema?
    - Quem são os atores (papéis) que interagem com que sistemas?

- O que os atores (papéis) podem fazer?
    - Através da diagramação, os casos de uso revelam onde cada ator pode interagir.




#### O Caso de Uso

Descreve as funcionalidades esperadas de um sistema em desenvolvimento.

O conjunto de todos os casos de uso descreve, em alto nível, todas as funcionalidades que os sistema deve prover.



#### Os Atores

Os atores interagem com o sistema quando:

1. Usam dos casos;
2. São usados pelos casos.

Atores não fazem parte do sistema, por isso ficam fora dos limites do sistema

Atores podem ser humanos ou não-humanos.



#### Tipos de Atores

- **Humano:** estudante, professor,...
- **Não-humano:** Servidor de e-mail.
- **Primário:** Principal beneficiário da execução do caso de uso
- **Secundário:** Não recebe nenhum benefício direto.
- **Ativo:** Inicia a execução do caso de uso. 
- **Passivo:** Propicia funcionalidades para a execução do caso de uso.



#### Associação entre Caso de Uso e Ator



##### Caso de Uso Base: 



###### A ---> B
- (A) Necessita que seja feita a execução de (B) antes que a própria (A) seja executada.



###### A <--- B
- (A) decide se vai ou não executar (B)




#### Dicas



##### Identificação de Atores

Quem usa os principais casos de uso?



##### Identificação de Casos de Uso
 
Quais são as principais tarefas que um ator precisa executar?

## **MIR Modelagem Informacional de Requisitos**



Foi proposta como uma especialização do modelo de Casos de Uso.

Em especial, o conceito de objetivo informacional substitui o de Caso de Uso, associando um nível de abstração específica aos objetivos considerados na modelagem.




### **Princípios da solução**

- **Focar nos Objetivos:** O objetivo é o nível mais alto de abstração de um processo. Logo é o mais adequado para uma primeira representação.

- **Atribuir níveis de abstração aos objetivos:** Corresponde a uma estratégia de aprticionamento funcional do sistema com base no conhecimento do problema.

- **Focar no detalhamento da informação:** Cada objetivo exprime uma mudança de estado que é concretizada pela execução do seu processo subjacente. O fluxo de informação que entra e sai do sistema é o foco do detalhe.



### **Objetivos Informacionais**

Quais eventos originados pelos atores externos levam o sistema a intervir atomicamentemente?



#### **Passos Necessários:**

1. **Definir** os **atores e objetivos informacioanis** numa tabela.
2. Fazer a **interface do sistema informacional** com:
    - **Descrição**;
    - **Propósito**;
    - **Frequência**;
3. **Dicionário** de itens elementares;
4. **Objetivos organizacionais**.
    - **Linhas temporais** com a sequência de objetivos informacionais. 

Sempre tentar ser o mais **simples e claro** o possível.

## **DW**



### **Informação Transacional vs Analítica**

- **Transacional / Operacional:** Armazenamento num banco de dados para o uso cotidiano.

- **Analítico:** Armazenamento feito para análise de dados e tomadas de decisões.




### **Conceitos de um DW**

O DW é um repositório estruturado de informação. O propósito é a extração de informações analíticas. O DW pode guardar informações detalhadas e temporais.



#### **Repositório Estruturados**
- O DW é um banco de dados contendo informações analíticas.



#### **Integrado**
- Se refere ao processo de juntar informações de vários bancos de dados diferentes em um único DW.



#### **Orientado por Objetivo**
- O DW é desenvolvido para um objetivo de negócio específico.



#### **Visão Ampla Empresarial**
- Refere-se ao fato de dar uma grande visão analítica da empresa como um todo.



#### **Histórico**
- Grande horizonte temporal de informações quando comparado com um banco de dados operacional.



#### **Coleta de Informações Analíticas**
- Somente para leitura dos usuários.



### **Componentes de um DW:**



#### **Sistemas de origem**
- São bases de dados operacionais e outros repositórios de dados.
- Pode-se incluir fontes internas e externas.



#### **Data Warehouse**
- Destino dos dados de origem
- Sempre lê os dados com uma certa frequência, oferecendo dados atualizados.



#### **Infraestrutura de ETL**

- A infraestrutura de ETL facilita a leitura de dados das bases oepracionais para os DW.



#### **Aplicações Front-End**

- Permitem o acesso indireto dos usuários aos dados.




### **Data Marts**

- Segue o mesmo princípio de um DW. Porém possui um escopo mais limitado.

## **Modelo Dimensional**

- Uma metodologia de design usada para banco de dados orientados por um objetivo
- Além de usar o básico de banco de dados, como chaves primárias, chaves estrangeiras, ... A modelagem dimensional distingue dois tipos de tabelas: 
    - **Dimensão**
    - **Fato**

### **Tabela Dimensional**

- Contém descrição do negócio, organização ou empresarial para qual o objetivo de análise sera orientado.


### **Tabela Fato**

- Contém chaves estrangeiras para as tabelas dimensionais.
- Pode conter valores de medidas para análises de negócio.

### **Esquema Estrela**

- O resultado fica parecendo uma estrela.

## **Data Lake**

- Representa uma abordagem mais ágil e flexível para a modelagem informacional
- Todos os dados devem ser armazenados e não há uma estrutura específica para eles

### **Big Data**

#### **Características**

- Conjuntos de dados em empresas e corporações que apresentam um **grande volume** de dados **diversificados** e em **rápido crescimento**. Têm as características de **não apresentarem metadados** e não são modelados formalmentes para a consulta e recuperação.

- Tipicamente **não-estruturado** ou **semi-estruturado**
    - Dados **não-estruturados**:
        - Sem estrutura explícita
        - Ex: documentos de texto
    - Dados **Semi-estruturados**:
        - Possui alguma estrutura (mas não completa)
        - Uma mensagem possui características de quem enviou, quem recebeu, horário,... Mas o corpo de texto não possuim uma estrutura definida

- A falta de metadados formais em conjuntos de Big Data se deve ao fato de que os repositórios de Big Data, ao contrário de DWs e BDs operacionais, não seguem uma estrutura pré-definida, ou seja, não seguem uma modelagem formal de bancos de dados.

#### **Os V's do Big Data**

- **Volume:**  
    - Grande volumetria de dados, onde apenas armazenar em um laptop comum é algo fora de questão




- **Variedade:**
    - Abundância de diferentes tipos de fontes de dados.
    


- **Velocidade:**
    - Alta velocidade de entrada de dados (Logs de wifi, chips em celulares com retornos do status atual, ...)

- **Veracidade:**
    - Problema de qualidade dos dados

- **Variabilidade:**
    - Possibilidade de interpretar os dados de diferentes maneiras
    

- **Valor:**
    - Utilidade e capacidade de base para ação dos dados 
    

- **Visualization:**
    - Necessidade de visualização elaborada dos dados

#### **Definição de Big Data**

- Big Data refere-se a grande volumes de conjuntos de dados diversificados e decrescimento rápido. Quando comparados com bancos de dados e DWs:


- 
    - São consideravelmente menos estruturados, com poucos ou nenhum metadados
    - Em gral, apresentam um maior Volume, Velocidade e Variedade
    - Têm uma maior probabilidade de apresentarem problemas de qualidade de dados (Veracidade)
    - Têm uma maio possibilidade de ter diferentes interpretações (Variabilidade)
    - Precisa de uma abordagem mai exploratória e experimental para gerar valor real nos dados, se houver

#### **MapReduce e Hadoop**

- Bancos de dados comuns não conseguem realizar consultas em grande volumes de dados de forma satisfatória

##### **MapReduce**

- Uma abordagem para lidar com Big Data
    - **Map:** mapeia cada registro em um par chave-valor
    - **Reduce:** reúne todos os registros com a mesma chave e gera um único registro de saída para cada chave



- **Hadoop**
    - Open-source MapReduce implementation

#### **DL vs DW** 

##### **Big Data Methods:**


- **Aumentar** a capacidade de analisar os dados que a empresa possui ou tem acesso

    - O Big Data não é um problema que deve ser abordado de forma separada e distinta.
    - A abordagem integrada, onde o Big Data anda de maõs dadas com outros projetos, é mais benéfica.

- **Não substitua** as abrodagens tradicioanis de bancos de dados  

##### **Data Warehouse**

- Armazena dados em uma estrutura de destino predefinida com metadados detalhados
    - Os usuários analisam e recuperam os dados
    

##### **Data Lake**

- Um grande pool de dadosem que o esquema e requisitos de dados não são definidos até que os dados sejam consultados
- O data lake armazena seus dados em formatos brutos, até que seja necessário 
    - Os usuários devem preparar seus dados antes de usá-los

### **Arquiteturas DL Zaloni e Inmon**

#### **Arquitetura de Zaloni**

- A arquitetura de Zaloni foca em questões operacionais, sobre como o Data Lake é gerenciado.
- 4 Zonas e um Sandbox. Onde cada zona ajudam a entender em qual estado os dados estão.

##### **Sandbox**

- Permite a análise exploratória sobre todas as zonas do Data Lake sem a necessidade de envolver a TI. Acessos são dados com base nos critérios da empresa

##### **Transiente**

- Camada temporária, usada apenas para a primeira carga e checagens de segurança e políticas da empresa (LGPD, GDPR).


##### **Zona Bruta**

- Após a checagem de qualidade dos dados e transformações de segurança, inicia-se a jornada do Data Lake.
- Aqui o dado é armazenado e catalogado de forma permanente.


##### **Zona Confiável**

- Os dados foram pré-processados e validados.
- Exemplo: Um mesmo cliente em vários sistemas aparece aqui como um único registro



##### **Zona Refinada**
- Os dados foram formatados para facilitar o uso geral  e/ou atender uma linha específica de negócio.

#### **Governança do Data Lake**

- O data lake pode atender casos ad hoc, mas sem governança dificilmente haverá confiança no data lake para a entrega de produtos essenciais de negócio.

-  **Qualidade e Consistência:**
    - O que é necessário para garantir que os dados tenham qualidade e consistência o suficiente para serem úteis aos usuários de negócio e cientistas de dados?

- **Políticas e Padrões:**
    - Quais são as políticas e padrões para ingestão, transformação e uso dos dados, e elas são observadas de forma uniforme em toda a organização?

- **Segurança e Conformidade:**
    - O acesso de dados confidenciais é limitado àqueles com a devida autorização?

- **Gestão do ciclo de vida do dado:**
    - Como gerenciaremos o ciclo de vida dos dados? Em que momento eles deverão ser arquivados em camadas de armazenamento mais baratas?

#### **Gerando Valor a partir do Data Lake**

- **Autonomia:**
    - Diferentes tipos de usuários querem acessar os dados em busca de algo diferente. Isso deve ocorrer sem a ajuda da TI.
- **Autoserviço:**
    - A experiência deve ser simples e útil
- **Todos os Públicos:**
    - Todo mundo deve poder ler o Data Lake, inclusive outros sistemas, como o DW.

#### **Plataforma de Autoserviço**

- O objetivo de um data lake é fornecer valor aos negócios atendendo aos usuários

#### **Controle de Acesso**

- Os data lakes precisam permitir que vários usuários de vários departamentos usem ferramenteas de que precisam e descubram insights através das relações entre os conjuntos de dados.

- É necessário ter políticas de segurança em relação aos acessos. Um usuário somente pode ver aquilo que lhe é permitido.

#### **Estratégia de Governança (crowdsourcing)**

- A estratégia se baseia em deixar os próprios usuários criarem novos modelos de dados a partir de conjuntos de dados existentes.

##### **Três Principais Tipos de Dados**

- **Dados Analógicos**
    - Dados de telemetria (GPS, logs, ...)
    - Volumosos e repetitivos.
    - tipicamente apenas outliers são alvo de interesse.

- **Dados de Aplicação**
    - Dados gerados a partir da execução de uma aplicação.
    - Uniformes.

- **Dados Textuais**
    - Não possui estrutura uniforme.
    - Pode assumir qualquer forma.

##### **As Lagoas do data lake (data pounds)**



- 1. Dados Brutos
    - 2. Dados Analógicos
    - 2. Dados de Aplicação
    - 2. Dados Textuais
        - 3. Dados de Arquivamento

- Cada data pound, exceto o de dados brutos, possui características em comum:
    - **Pound Descriptor**
    - **Pound Target**
    - **Pound Data**
    - **Pound Metadata**
    - **Pound Metaprocess**
    - **Pound Transformation Criteria**

#### **Zaloni vs Inmon**

- **Zaloni:** 
    - Leva em conta que os dados passam por estágios (zonas) de refinamento

- **Inmon:**
    - Leva em conta os tipos de dados e prevê uma segregação arquitetônica de acordo

- Ambas as abordagens levam em conta a governança, metadados e linhagem de dados. Assim evintando que o data lake se transforme em um pântano de dados.
