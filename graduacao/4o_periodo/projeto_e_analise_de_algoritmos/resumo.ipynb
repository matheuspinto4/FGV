### **Análise da Complexidade**

Podemos definir a ocmplexidade de um determinado algoritmo através de uma função como:

$$
T(n) = \text{número de instruções} \\
\text{executadas considerando} \\ 
\text{a intância que exercita o} \\
\text{pior caso}
$$



### **Notação Assintótica**



#### **Motivação**

Determinar **com precisão** a função que descreve a complexidade de um algoritmo pode ser uma tarefa difícil e muito trabalhosa

- A **ordem assintótica de crescimento** de $T(n)$ é definida encontrando as funções que descrevem os **limites superior e inferior** de $T(n)$



#### **Notação O:**

**$T(n)$** pertence ao conjunto **$O(f(n))$** se existirem constantes positivas **$c$** e **$n_0$** tais que:

**$$
T(n) \le c f(n) \text{ }\forall \text{ }
 n \ge n_0$$**



#### **Notação $\Omega$:**


**$T(n)$** pertence ao conjunto **$\Omega(f(n))$** se existirem constantes positivas **$c$** e **$n_0$** tais que:

**$$
c f(n) \le T(n)\text{ }\forall \text{ }
 n \ge n_0$$**




#### **Notação $\Theta$:**

$T(n)$ pertence ao conjunto $\Theta(f(n))$ se $T(n)$ for $O(f(n))$ e $\Omega(f(n))$

#### **Substituição**
O método se baseia em dar um hipótese de indução e tentar provar ela.

$$
T(n) = T(n/2) + n \\
T(n) \le cn\log(n) \text{ hipótese} \\
T(n/2) \le c\frac{n}{2}\log(\frac{n}{2}) \\
T(n) \le cn\log(n)
$$



#### **Árvore de Recursão**

No método se somam sa complexidades de cada nó da árvore.




#### **Método da Iteração**

Se expande a expressão e tenta resolver.



#### **Método Mestre**
**$$
T(n) = aT(\frac{n}{b}) + f(n)$$**


Temos 3 casos:

1. Se $f(n) = O(n^{\log_b{a-\epsilon}})$ for some constant $\epsilon > 0$, then $T(n) = \Theta(n^{\log_ba})$

\

2. Se $f(n) = \Theta(n^{\log _b a})$, then $T(n) = \Theta(n^{\log_ba}\lg n)$

\

3. Se $f(n) = \Omega(n^{\log_ba + \epsilon})$ for some constant $\epsilon > 0$, and if $af(n/b) \le cf(n)$ for some constant $c < 1$ and all sufficiently large $n$, then $T(n) = \Theta(f(n))$




## **Algoritmos de Busca**


- A comparação é feita por uma função ```compara(a, b)```


### **Busca em Vetor Ordenado**

Complexidade: $O(\log n)$
```python
def busca_binaria(lista, alvo, compara):
    baixo, alto = 0, len(lista) - 1
    while baixo <= alto:
        meio = (baixo + alto) // 2
        comp = compara(lista[meio], alvo)
        if comp == 0:
            return meio
        elif comp < 0:
            baixo = meio + 1
        else:
            alto = meio - 1
    return -1

```

## **Ordenação**

- Um algoritmo de ordenação é considerado **estável** se não alterar a posição relativa de elementos com o mesmo valor.

### **Bubble Sort**

- A solução consiste em inverter valores de posições adjacentes sempre que $v[i+1] < v[i]$

- $\Theta(n^2)$

### **Selection Sort**

- Dada uma posição $i$, busca-se encontrar a posição $j$ com o menor valor dentre as posições $[i..n-1]$
- $\Theta(n^2)$

### **Insertion Sort**

- Percorre-se o vetor, posicionando o valor de índice $j$ na parte devida nos indices $[0..j-1]$ para que ela esteja ordenada

- $\Theta(n^2)$

### **Merge Sort**

- Problema: dadas duas sequências A e B ordenadas produza uma sequência R ordenada contendo os elementos de ambas as sequências:

```python
def merge(esq, dir):
    res, i, j = [], 0, 0
    while i < len(esq) and j < len(dir):
        if esq[i] <= dir[j]:
            res.append(esq[i]); i += 1
        else:
            res.append(dir[j]); j += 1
    return res + esq[i:] + dir[j:]

```

- O algoritmo de mergesort usa este fato, dividindo a lista em duas filhas e ordenando estas, após é feita a ordenação destas listas. Este é um processo recursivo.

```python
def merge_sort(lst):
    if len(lst) <= 1: return lst
    meio = len(lst) // 2
    return merge(merge_sort(lst[:meio]), merge_sort(lst[meio:]))

```

- A complexidade é $T(n) = T(\frac{n}{2}) + T(\frac{n}{2}) + n$ se $n>1$. 
- Portanto, fica em $\Theta(n\log n)$

### **Quicksort**

- Dada uma lista, ache a posição final de um elemento pivô:

```python
def particionar(lst, baixo, alto):
    pivo, i = lst[alto], baixo - 1
    for j in range(baixo, alto):
        if lst[j] <= pivo:
            i += 1
            lst[i], lst[j] = lst[j], lst[i]
    lst[i+1], lst[alto] = lst[alto], lst[i+1]
    return i + 1
```

- Algoritmo **QuickSort**:
```python
def quick_sort(lst, baixo=0, alto=None):
    if alto is None: 
        alto = len(lst) - 1
    if baixo < alto:
        pi = particionar(lst, baixo, alto)
        quick_sort(lst, baixo, pi - 1)
        quick_sort(lst, pi + 1, alto)
    return lst
```

- Complexidade média: $\Theta(n \log n)$
- Complexidade no pior caso: $\Theta(n^2)$

### **MOM**

```python
def select_kth(arr, k):
    if len(arr) <= 10: return sorted(arr)[k-1]

    medians = [sorted(arr[i:i+5])[len(arr[i:i+5])//2] for i in range(0, len(arr), 5)]
    pivot = select_kth(medians, len(medians)//2)

    less = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    greater = [x for x in arr if x > pivot]

    if k < len(less): return select_kth(less, k)
    if k < len(less) + len(equal): return pivot
    return select_kth(greater, k - len(less) - len(equal))
```

## **Introdução à Grafos**

### **Representação**

- **tamanho:** $|V| + |E|$
- **extremos de uma aresta:** $e = (v_i, v_j)$ 
- **aresta incidente:** $e$ é incidente em $v_i$ e  $v_J$
- **adjacência:** $v_i$ e $v_j$ são adjacentes

### **Características**

##### **Denso**

- Um grafo é denso se o seu tamanho for proporcional ao quadrado do número de vértices
    - **$(|V| + |E|) \propto |V|^2$**

#### **Esparso**

- Um grafo é esparso se o seu tamanho for proporcional ao número de vértices
    - **$(|V| + |E|) \propto |V|$** 

### Árvore

- Um grafo é uma árvore se:
    - Possui exatamente $|V| - 1$ arestas
    - A remoção de qualquer aresta torna o grafo desconexo
    - Para todo par de vértices existe um único caminho 

- Um grafo é uma floresta se for um grafo acíclico

## **Busca em Grafos:**

### **Caminho**


- Um **caminho** $P$ em $G = (V, E)$ consiste em uma sequência de $n$ vértices, finita e não vazia, tal que $v_{i + 1}$ é adjacênte a $v_i$ para todo $i < n$
- Propriedades:
    - **Simples:** não possui vértices repetidos.
    - **Fechado (ciclo):** se $v1 = v_n$
    - **Comprimento:** número de arestas nele 

- Para checar se um caminho existe em um grafo:
    - Percorrer o caminho e para cada par de vértice nele ver se a aresta existe no grafo

### **Grafos Topológicos**


- É um grafo que admite uma ordenação dos vértices de forma que para toda aresta **$(v_i, v_j)$** temos que **$i < j$**

- Grafos topológicos não apresentam ciclos


-    - A ideia para checar se tem uma ordem topológica consistem em ver se é possível eliminar todos os nós do grafo.
-    - Para eliminar um nó de um grafo, é preciso retirar a dependência dele de todos os seus antecessores
-    - Se utiliza a in-degree para realizar a ordem

### **Florestas DFS**

- Uma árvore radicada é um grafo topológico sem vértices com grau de entrada maior do que 1.
- O algoritmo dfs gera uma floresta DFS

- **preOrder[v]** é quando o vértice foi descoberto
- **postOrder[v]** é quando o vértice é exaurido (Todas as arestas de saídas foram examinadas)

### **Intervalo de Vida**

- O **intervalo de vida** (lifespan) de um vértice no contexto da busca ocorre entre o momento que ele é descoberto e o momento em que ele é exaurido.
    - Definido por **preOrder[v]** e **postOrder[v]**
- Tomar cuidado, pois **preOrder[v]** e **postOrder[v]** são numerações independentes. 

### **Ancestrais, descendentes e primos**

- Dados dois vértices $v_1$ e $v_2$ de uma floresta radicada produzida por uma execução DFS.

- Primos:
    - $v_1$ é primo mais **velho** de $v_2$ se
        - preOrder[$v_1$] $<$ postOrder[$v_2$]
    - $v_1$ é primo mais **novo** de $v_2$ se
        - preOrder[$v_1$] $>$ postOrder[$v_2$]

- Arestas **$(v_i, v_j)$** que **não pertencem** à floresta DFS podem ser classificadas de acordo com o grau de parentesco.
    - **Retorno:**
        - $v_j$ é **ancestral** de $v_i$
    - **Avanço:**
        - $v_j$ é **descendente** de $v_i$
    - **Cruzada:**
        - $v_j$ é **primo** de $v_i$
        - Arestas cruzadas são sempre de um primo mais novo para um mais velho.

### **Ciclos**

- Um grafo é acíclico se e somente se não possuir uma numeração topológica


- Para saber se um grafo é acíclico, podemos realizar uma DFS nele e checar se existe uma aresta de retorno, comparando os intervalos de vida para cada vértice.

### **BFS**

- O grafo é percorrido em níveis 

```python
def bfs(grafo, no_inicial):
    fila = deque([no_inicial])
    visitados = {no_inicial}
    ordem_visita = []
    while fila:
        no_atual = fila.popleft()
        ordem_visita.append(no_atual)
        for vizinho in grafo.get(no_atual, []):
            if vizinho not in visitados:
                visitados.add(vizinho)
                fila.append(vizinho)
    return ordem_visita
```

## **Menor Caminho em Grafos**

### **Busca pelo Caminho mais Curto**

- A comparação d[$v_i$] $+$ $1$ $\le$ d[$v_j$] e a eventual atualização no vetor de distâncias é conhecida como **operação de relaxamento**. 

- Uma aresta está **relaxada** se d[$v_j$] - d[$v_i$] $\le 1$
- Uma aresta está **tensa** se d[$v_j$] - d[$v_i$] $\gt 1$

```python
def bfs_caminho_mais_curto(grafo, inicio):
    distancias = {no: float('inf') for no in grafo}
    predecessores = {no: None for no in grafo}
    distancias[inicio] = 0
    fila = deque([inicio])
    while fila:
        u = fila.popleft()
        for v in grafo.get(u, []):
            if distancias[v] == float('inf'):
                distancias[v] = distancias[u] + 1
                predecessores[v] = u
                fila.append(v)
    return distancias, predecessores
```

### **Busca pelo Caminho mais Barato**

- Para encontrar o **caminho mais barato** entre $v_i$ e $v_j$ é inevitável encontrar todos os caminhos que iniciam em $v_i$
- A árvore produzida por essa busca é chamada de CPT.
- Ela só existe se não houver ciclo negativo 

### **Dijkstra**

- Encontra caminhos mais baratos em um grafo com arestas positivas



- A solução consiste em crescer uma árvore radicada a partir do vértice inicial $v_0$ até que ela seja uma árvore geradora do subgrafo induzido a partir de $v_0$

- **Franja:** 
    - A franja de uma árvore radicada $T$ com raiz em $v_0$ é o conjunto de arestas $(v_i, v_j)$ que possuem $v_i$ em $T$ e $v_j$ fora de $T$
    - A franja pode ser vista como o grau de saída do conjunto de vértices em $T$.


- Ideia geral:
    - Insira o vértice $v_0$ em $T$ e defina $d[v_0] = 0$
    - Enquanto a franja não estiver vazia:
        - Escolha o vértice $v_k$ que apresente a menor distância.
        - Aplique a operação de relaxamento em todas as arestas de $v_k$ que pertencem à franja.
        - Insira o vértice $v_k$ em $T$

```python
def dijkstra(grafo, inicio):    
    distancias = {vertice: float('inf') for vertice in grafo}
    distancias[inicio] = 0
    fila_prioridade = [(0, inicio)]  
    while fila_prioridade:
        d_u, u = heapq.heappop(fila_prioridade)
        if d_u > distancias[u]:
            continue
        for v, peso in grafo.get(u, {}).items():
            nova_distancia = distancias[u] + peso
            if nova_distancia < distancias[v]:
                distancias[v] = nova_distancia
                heapq.heappush(fila_prioridade, (nova_distancia, v))
    return distancias
```

- **$O((V + E)\log(V))$**

- Se todos os vértices forem acessiveis a partir de $v_0$: **$O(E\log(V))$**

### **Bellman-Ford**

```python
def bellman_ford(G, n):
    d = {u: float("inf") for u in G}
    d[1] = 0
    for _ in range(n - 1):
        #  Para cada aresta, realizar o relaxamento
        for u in G:
            for v, peso in G[u].items():
                if d[u] + peso < d[v]:
                    d[v] = d[u] + peso
    
    for u in G:
        for v, peso in G[u].items():
            if d[u] + peso < d[v]:
                return "Ciclo Negativo"

    return d
```

## **Árvore Geradora Mínima**

- Uma árvore geradora de um grafo G é o subgrafo T que não possua ciclos e que contenha todos os vértices de G

- Um **corte** é um conjunto de arestas que conecta duas partes de um grafo
- Dado um conjunto de vértices $A$, as arestas que possuem uma ponta em $A$ e outra em $\bar{A}$ representam um corte
- $A$ e $\bar{A}$ são as margens do corte.

```python

def prim(G):
    custo_total = 0
    mst = set()
    d = {v : float("inf") for v in G.keys()}
    d[1] = 0
    T = [(0, 1)]  #  (peso, vertice)

    while T:
        peso, u = T.heapq.heappop(T)

        if u in mst: continue
        mst.add(u)
        custo_total += peso
        for v, peso_uv in G.get(u, {}).items():
            if v not in mst and peso_uv < d[v]:
                d[v] = peso_uv
                heapq.heappush(T, (peso_uv, v))
    
    return custo_total

```

## **P x NP**

### **Problemas de Decisão**

#### **P**

- P é a classe de problemas de decisão que podem ser resolvidos em tempo polinomial

#### **NP**

- Classe de problemas de decisão.
- Ainda não acharam uma solução em tempo polinomial para resolver
- Qualquer solução pode ser checada em tempo polinomial 

#### **NP-completo**

- Um problema B é NP-completo se:
    - B $\in$ NP
    - $\forall A \in NP$ $A \propto B$  (A é redutível ao B)

- Se A e B em NP-completo, de modo que $A \propto B$:
    - Se $B \in P$, então $A \in P$
    - Se $A \in NP-completo$, então $B \in NP-completo$

### **Problemas de Otimização**

#### **NP-difícil**

- Não necessariamente NP-dificil precisa ser NP
