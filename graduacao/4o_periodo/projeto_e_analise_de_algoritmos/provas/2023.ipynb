```python
def q2(A):
    inicio = 0
    fim = len(A) - 1
    val = None
    while(inicio <= fim):
        meio = (inicio + fim) // 2
        if A[meio] == meio:
            val = meio
        if A[meio] > meio: fim = meio - 1
        else: inicio = meio + 1 
    
    return val
```

```python
def q4(A, k):

    def mom(A, k):
        if len(A) <= 10: return sorted(A)[k-1]

        medians = [sorted(A[i:i+5])[len(A[i:i+5])//2] for i in range(0, len(A), 5)]
        pivot = mom(medians, k)

        less = [x for x in A if x < k]
        equal = [x for x in A if x == k]
        greater = [x for x in A if x > k]

        if k < len(less): mom(less, k)
        elif k < len(less) + len(equal): return pivot
        return mom(greater, k - len(less) - len(equal)) 

    mediana = mom(A, k=len(A) // 2)
    lista_modulos = [abs(mediana - x) for x in A]
    valor_limiar = mom(lista_modulos, k=k)
    l =  []
    l2 =  []
    for i,x in enumerate(lista_modulos):
        if x <= valor_limiar:
            l.append(A[i])
            l2.append(x)

    if len(l) > k:
        l.remove(A[max(l2)])

    return l
            
A = [1,29,30,4,50,6,9]

q4(A, k=3)

```
