```python
def q4_1(A):
    # O(n^3)
    melhor = float("-inf")
    i_melhor, j_melhor = 0,0
    for i in range(len(A)):
        for j in range(i+1, len(A)):
            s = sum(A[i:j+1])
            # print(i, j, s)
            ruim = False
            for k in range(j + 1, len(A)):
                s2 = sum(A[i:k])
                if s2 > s: ruim=True
            if ruim : continue
            if s > melhor:
                melhor = s
                i_melhor, j_melhor = i, j

    return (melhor, i_melhor, j_melhor)

def q4_2(A):
    melhor = float("-inf")
    i_melhor, j_melhor = 0,0
    for i in range(len(A)):
        for j in range(i+1, len(A)):
            s = sum(A[i:j+1])
            # print(i, j, s)
            if s > melhor:
                melhor = s
                i_melhor, j_melhor = i, j

    return (melhor, i_melhor, j_melhor)

def q4_3(A):
    melhor = float("-inf")
    i_melhor, j_melhor = 0,0
    sum_acumulada = A[0]
    i = 0
    for j in range(1, len(A)):
        if i == j: continue
        if A[j] + sum_acumulada >= 0: 
            sum_acumulada += A[j]
            if sum_acumulada >= melhor and i != j:
                i_melhor, j_melhor = i, j
                melhor = sum_acumulada
        else:
            if sum_acumulada >= melhor and i != j:
                i_melhor, j_melhor = i, j
                melhor = sum_acumulada
            sum_acumulada = 0
            i = j + 1

    
    print(melhor, i_melhor, j_melhor)
    
                

A = [0,-50,-6000, 100, -900]
q4_3(A)
```
