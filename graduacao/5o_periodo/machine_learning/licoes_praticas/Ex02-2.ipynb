# Cross-validation

**1.** O código abaixo carrega o banco de dados *California housing*. Divida o banco de dados em treino, teste e validação. Use o conjunto de validação para escolher o coeficiente de regularização $c$ para um modelo de regressão linear com penalização $L_2$. Use a fórmula analítica para estimar os pesos do modelo de regressão. Plote os MSE no conjunto de trieno e validação em função de $c$. Comente o resultado. Avalie a performance do modelo ótimo no conjunto de teste e também comente.

```python
#  --  Dividindo o banco de dados em treino, teste e validação:  --  #

SEED = 42
np.random.seed(SEED)

X, y = fetch_california_housing(return_X_y=True)
```

```python
#  Separando em treino, validação e teste
X_train_validation, X_test, y_train_validation, y_test = train_test_split(
    X, y, test_size=0.25, random_state=SEED
)
X_train, X_validation, y_train, y_validation = train_test_split(
    X_train_validation, y_train_validation, test_size=0.25, random_state=SEED
)

#  Padronizando os dados
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_validation = scaler.transform(X_validation)
X_test = scaler.transform(X_test)

#  Adicionando o Bias
X_train = np.c_[np.ones((X_train.shape[0], 1)), X_train]
X_validation = np.c_[np.ones((X_validation.shape[0], 1)), X_validation]
X_test = np.c_[np.ones((X_test.shape[0], 1)), X_test]
```

```python
#  -- Estimando os parâmetros e computando o erro -- 

def estimar_pesos_L2(X, y, c):
    n = X.shape[1]
    I = np.eye(n) 
    I[0, 0] = 0 
    theta_hat = np.linalg.inv(X.T@X + I*c) @ (X.T @ y)
    
    return theta_hat

def MSE(y_prev, y_real):
    return np.mean( (y_prev - y_real) ** 2)

c_list = np.logspace(2, 6, 10000)
mse_list_validation = []
mse_list_train = []
for c in c_list:
    theta_hat = estimar_pesos_L2(X_train, y_train, c)
    y_prev_validation = X_validation @ theta_hat
    y_prev_train = X_train @ theta_hat
    mse_list_validation.append(MSE(y_prev_validation, y_validation))
    mse_list_train.append(MSE(y_prev_train, y_train))
```

```python
#  -- PLotando o erro pelo valor de c --  #

argmin = np.argmin(mse_list_validation)
c_otimo = c_list[argmin]
print(f"min = {mse_list_validation[argmin]} for c = {c_otimo}" )
plt.plot(c_list, mse_list_validation, label = "Validação")
plt.plot(c_list, mse_list_train, label = "Treino")
plt.axvline(x = c_otimo, linestyle="--", color="red", label="c_ótimo", alpha=0.6)
plt.yscale("log")
plt.xscale("log")
plt.xlabel("c")
plt.ylabel("MSE")
plt.legend()
plt.show()
```

```python
#  -- Testando o desempenho para o c ótimo -- #

X_train_validation, X_test, y_train_validation, y_test = train_test_split(
    X, y, test_size=0.25, random_state=SEED
)

X_train_validation = scaler.fit_transform(X_train_validation)
X_test = scaler.transform(X_test)

X_train_validation = np.c_[np.ones((X_train_validation.shape[0], 1)), X_train_validation]
X_test = np.c_[np.ones((X_test.shape[0], 1)), X_test]

theta_hat = estimar_pesos_L2(X_train_validation, y_train_validation, c_otimo)
y_prev_test = X_test @ theta_hat
print(f"MSE para o teste: {MSE(y_prev_test, y_test)}")
```

```python
class LinearRegression():
    def __init__(self) -> None:
        self.scaler = StandardScaler()
        pass

    def fit(self, X, y, c=0):
        X_ = self.scaler.fit_transform(X)
        X_ = np.c_[np.ones((X_.shape[0], 1)), X_]
        n = X_.shape[1]
        I = np.eye(n) 
        I[0, 0] = 0 
        theta_hat = np.linalg.inv(X_.T@X_ + I*c) @ (X_.T @ y)
        
        self.theta_hat = theta_hat
        return theta_hat

    def predict(self, X):
        X_ = self.scaler.transform(X)
        X_ = np.c_[np.ones((X_.shape[0], 1)), X_]

        y_prev = X_ @ self.theta_hat
        
        return y_prev
    

def cross_validation(X, y, K, model, scoring=MSE, SEED=42):
    n = len(X)
    X_shuffled, y_shuffled = shuffle(X, y, random_state=SEED)
    scoring_list = []

    for k in range(1, K+1):
        i1, i2 = int((k-1)/K * n), int((k)/K * n)  #  Achando os indices das divisões para o teste
        
        X_test, y_test = X_shuffled[i1:i2], y_shuffled[i1:i2]
        X_train = np.concatenate((X_shuffled[:i1], X_shuffled[i2:]), axis=0)
        y_train = np.concatenate((y_shuffled[:i1], y_shuffled[i2:]), axis=0)

        model.fit(X_train, y_train)
        y_test_pred = model.predict(X_test)
        scoring_list.append(scoring(y_test_pred, y_test))
    
    return np.mean(scoring_list)


def nested_cross_validation(X, y, K, model, hyperparams=[], scoring=MSE, SEED=42):
    n = len(X)
    X_shuffled, y_shuffled = shuffle(X, y, random_state=SEED)
    scoring_test_list = []

    dic = {}
    for hyper in hyperparams:
        dic[hyper] = 0

    for k in range(1, K+1):
        i1, i2 = int((k-1)/K * n), int((k)/K * n)  #  Achando os indices das divisões para o teste
        
        X_test, y_test = X_shuffled[i1:i2], y_shuffled[i1:i2]
        X_train_vali = np.concatenate((X_shuffled[:i1], X_shuffled[i2:]), axis=0)
        y_train_vali = np.concatenate((y_shuffled[:i1], y_shuffled[i2:]), axis=0)


        n_train_vali = ((K - 1) / K ) * n
        hyper_scoring_list = []
        for hyper in hyperparams:
            scoring_vali_list = []
            for k_ in range(1, K):
                i1_, i2_ = int((k_-1)/(K - 1) * n_train_vali), int((k_)/(K - 1) * n_train_vali)  #  Achando os indices das divisões para a validação
                X_vali, y_vali = X_train_vali[i1_:i2_], y_train_vali[i1_: i2_]
                X_train = np.concatenate((X_train_vali[:i1_], X_train_vali[i2_:]), axis=0)
                y_train = np.concatenate((y_train_vali[:i1_], y_train_vali[i2_:]), axis=0)
                
                if type(model).__name__ == "LinearRegression":            
                    model.fit(X_train, y_train, c=hyper)
                
                if type(model).__name__ ==  'KNeighborsRegressor':
                    model.n_neighbors = hyper
                    model.fit(X_train, y_train)
                #  Treinando e validando os hiperparâmetros

                y_vali_pred = model.predict(X_vali)
                scoring_vali_list.append(scoring(y_vali_pred, y_vali))

            hyper_scoring_list.append(np.mean(scoring_vali_list))
            
        #  Achando o melhor hyperparâmetro para o conjunto
        hyper_otimo = hyperparams[np.argmin(hyper_scoring_list)]
        # print(f"Melhor Hyperparâmetro para {type(model)}= {hyper_otimo}")
        dic[hyper_otimo] += 1

        #  Treinando para o c_otimo
        if type(model).__name__ == 'LinearRegression':            
            model.fit(X_train_vali, y_train_vali, c=hyper_otimo)
                
        if type(model).__name__ ==  'KNeighborsRegressor':
            model.n_neighbors = hyper_otimo
            model.fit(X_train_vali, y_train_vali)

        y_test_pred = model.predict(X_test)
        scoring_test_list.append(scoring(y_test_pred, y_test))

    plt.bar([str(k) for k in dic.keys()], dic.values(), color='skyblue', edgecolor='black')    
    plt.title(f"Frequência dos Hiperparâmetros Ótimos ({type(model).__name__})")
    plt.xlabel("Hiperparâmetro")
    plt.ylabel("Vezes Escolhido")
    plt.show()
    
    return np.mean(scoring_test_list)
```

```python
lr = LinearRegression()
knn = KNeighborsRegressor()

lr_scoring = nested_cross_validation(X, y, K=5, hyperparams=[0, 1, 10, 100], scoring=MSE, model=lr)
knn_scoring = nested_cross_validation(X, y, K=5, hyperparams=[1,2,3,4,5], model=knn, scoring=MSE)

print(f"Linear Regression Score: {lr_scoring}")
print(f"KNN Score: {knn_scoring}")

print(f"Melhor modelo {'LinearRegression' if lr_scoring <= knn_scoring else 'KNN'}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
import seaborn as sns
import io
import base64

import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import random_split, DataLoader
import tqdm
```

```python
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
print(device)

# The batch size
batch_size = 2048

transform_train = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])
transform_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)),
])

train_set = torchvision.datasets.CIFAR10(
    root="./data", train=True, download=True,
    transform=transform_train,
)
test_set = torchvision.datasets.CIFAR10(
    root="./data", train=False, download=True,
    transform=transform_test
)

# -- Criando o train, val e o test loader:
porcentagem_treino = 0.8
porcentagem_val = 0.2

gerador = torch.Generator().manual_seed(42)

treino_split, val_split = random_split(
    train_set,
    [porcentagem_treino, porcentagem_val],
    generator=gerador
)

train_val_loader = DataLoader(
    train_set, batch_size=batch_size, shuffle=True,
    num_workers=2
)

train_loader = DataLoader(
    treino_split, batch_size=batch_size, shuffle=True,
    num_workers=2
)
val_loader = DataLoader(
    val_split, batch_size=batch_size, shuffle=False,
    num_workers=2
)
test_loader = DataLoader(
    test_set, batch_size=batch_size, shuffle=False,
    num_workers=2
)

classes = (
    'plane', 'car', 'bird', 'cat', 'deer',
    'dog', 'frog', 'horse', 'ship', 'truck'
)

n_img = 8

f, ax = plt.subplots(2, 4, figsize=(9,5))
dataiter = iter(test_loader)
images, labels = next(dataiter)

for i in range(n_img):
    npimg = images[i]/2 + 0.5
    npimg = npimg.numpy()
    ax[i%2,i//2].imshow(np.transpose(npimg, (1, 2, 0)))
    ax[i%2,i//2].set_title(classes[labels[i]])

```

```python
from sklearn.model_selection import ParameterGrid

class CNN(nn.Module):
    def __init__(self, params):
        super().__init__()
        self.conv1 = nn.Conv2d(3, params['filtros_conv1'], params['kernel_size'])
        self.pool = nn.MaxPool2d(2, 2)
        self.conv2 = nn.Conv2d(params['filtros_conv1'], params['filtros_conv2'], params['kernel_size'])
        self.fc1 = nn.LazyLinear(params['num_neuronios_fc1'])
        self.fc2 = nn.Linear(params['num_neuronios_fc1'], params['num_neuronios_fc2'])
        self.fc3 = nn.Linear(params['num_neuronios_fc2'], 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = torch.flatten(x, 1) # flatten all dimensions except batch
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

```python
#  -- configurações de hiperparâmetros para testar

grid_de_parametros = {
    'learning_rate': [0.001, 0.0001],
    'filtros_conv1': [16, 32],
    'batch_size': [128, 256],
    'filtros_conv2': [32],
    'fator_scheduler': [0.5],
    'patience_scheduler': [3],
    'num_early_stopping': [10],
    'kernel_size': [3],
    'num_neuronios_fc1': [256],
    'num_neuronios_fc2': [128],
}
combinacoes = ParameterGrid(grid_de_parametros)
```

```python
#  -- Achando a melhor configuração para os hiperparâmetros
num_epochs = 1
best_global_loss = float("inf")
best_params = None
best_state = None
total_configs = len(combinacoes)
for i, params in enumerate(combinacoes, 1):
    configs_restantes = total_configs - i
    print(f"\n========================================================")
    print(f"Testando configuração {i} de {total_configs} (Restam: {configs_restantes})")
    print(f"Parâmetros: {params}")
    print(f"========================================================")
    cnn = CNN(params=params).to(device)

    criterion = nn.CrossEntropyLoss()
    optimizer_cnn = torch.optim.Adam(cnn.parameters(), lr=params['learning_rate'])    

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer_cnn,
        mode='min',
        factor=params['fator_scheduler'],
        patience=params['patience_scheduler']
    )
    
    # -- Mecânica do early stopping
    contador_earling_stopping = params['num_early_stopping']
    min_avg_loss = float("inf")
    for epoch in range(num_epochs):  
        # -- Treinando o modelo -- #
        cnn.train()
        running_loss = 0.0
        for data in train_loader:
            inputs, labels = data
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer_cnn.zero_grad()
            outputs = cnn(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer_cnn.step()

            running_loss += loss.item()
        
        # -- Validando o modelo -- #
        cnn.eval()
        val_loss = 0.0
        with torch.no_grad():
            for data in val_loader:
                inputs, labels = data
                inputs, labels = inputs.to(device), labels.to(device)

                outputs = cnn(inputs)
                loss = criterion(outputs, labels)
                val_loss += loss.item()
        
        avg_val_loss = val_loss / len(val_loader)

        scheduler.step(avg_val_loss)
        current_lr = optimizer_cnn.param_groups[0]['lr']
        print(f"Época {epoch+1} | Loss Treino: {running_loss/len(train_loader):.4f} | Loss Val: {avg_val_loss:.4f} | LR atual: {current_lr}")

        #  -- Mecânica do Early Stopping
        if avg_val_loss < min_avg_loss:
            min_avg_loss = avg_val_loss
            contador_earling_stopping = 0
        else:
            contador_earling_stopping += 1
        if contador_earling_stopping == params: break
    
    #  -- Salvando a melhor configuração
    if min_avg_loss < best_global_loss: 
        best_global_loss = min_avg_loss
        best_params = params
        best_state = cnn.state_dict()
```

```python
#  -- Treinando com o modelo ótimo
cnn = CNN(params=best_params).to(device)

criterion = nn.CrossEntropyLoss()
optimizer_cnn = torch.optim.Adam(cnn.parameters(), lr=best_params['learning_rate'])    

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_cnn,
    mode='min',
    factor=best_params['fator_scheduler'],
    patience=best_params['patience_scheduler']
)
    
loss_list = []
loss_por_epoca = []
for epoch in range(num_epochs * 10):  
    print(f"Época {epoch}")
    # -- Treinando o modelo -- #
    cnn.train()
    running_loss = 0.0
    for data in train_val_loader:
        inputs, labels = data
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer_cnn.zero_grad()
        outputs = cnn(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer_cnn.step()

        running_loss += loss.item()
        loss_list.append(loss.item())

    media_epoca = running_loss / len(train_val_loader)
    
    # E salvamos apenas 1 valor por época
    loss_por_epoca.append(media_epoca)
```

```python

plt.figure(figsize=(8, 5))
plt.plot(loss_por_epoca, marker='o', color='tab:blue', linewidth=2, label='Treino (Train + Val)')

plt.title(f"Curva de Aprendizado - Modelo Ótimo", fontsize=14)
plt.xlabel("Épocas", fontsize=12)
plt.ylabel("Erro (Loss)", fontsize=12)

# Adicionando um grid para facilitar a leitura e uma legenda
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#  -- Desempenho no conjunto de Teste -- #
cnn.eval()

loss = 0.0
for data in test_loader:
    inputs, labels = data
    inputs, labels = inputs.to(device), labels.to(device)
    outputs = cnn(inputs)
    loss += criterion(outputs, labels)

print(f"Perda no conjunto de teste: {loss / len(test_loader)}")

loss = 0.0
for data in train_val_loader:
    inputs, labels = data
    inputs, labels = inputs.to(device), labels.to(device)
    outputs = cnn(inputs)
    loss += criterion(outputs, labels)

print(f"Perda no conjunto de treino_val: {loss / len(train_val_loader)}")

gridX, gridY = 4, 4
n_img = gridX * gridY

f, ax = plt.subplots(gridX, gridY, figsize=(gridX * 3, gridY * 3.5))
dataiter = iter(test_loader)
images, labels = next(dataiter)

with torch.no_grad():
    for i in range(n_img):
        npimg = images[i]/2 + 0.5
        imagem = images[i].unsqueeze(0).to(device)
        
        _, saida = torch.max(cnn(imagem),1)

        npimg = npimg.numpy()
        ax[i%gridX,i//gridY].imshow(np.transpose(npimg, (1, 2, 0)))
        ax[i%gridX,i//gridY].set_title(f"Real: {classes[labels[i]]}, \n Previsto: {classes[saida]}")
        ax[i%gridX,i//gridY].axis("off")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
import torch
import torch.nn.functional as F
from torch.autograd.functional import hessian
from torch.distributions.multivariate_normal import MultivariateNormal
import seaborn as sns
import io
import base64
```

**1.** Durante a aula, discutimos como construir uma priori GP e o formato da posteriori preditiva para problemas de regressão com verossimilhança Gaussiana (com média definida pelo GP). O código abaixo cria um GP com kernel exponencial quadrático, mostra a priori preditiva e a posteriori preditiva. Experimente com o código e comente a influência de ambos os parâmetros do kernel exponencial quadrático, tanto na priori preditiva quanto na posteriori preditiva. Nos gráficos gerados, os pontos vermelhos são observações, as curvas sólidas azuis são a médias das preditivas e o sombreado denota +- um desvio padrão.

```python
SEED = 42
np.random.seed(SEED)

s2 = 1e-04 # variância observacional

def rbf_kernel(x1, x2, gamma=10.0, c=1.0):
    assert(gamma>0)
    assert(c>0)
    return (-gamma*(torch.cdist(x1, x2)**2)).exp()*c

x = torch.linspace(-1, 1, 100)[:, None]

K = rbf_kernel(x, x) + torch.eye(x.shape[0])*s2
mu = torch.zeros_like(x)

fig, axs = plt.subplots(1, 2, figsize=(9, 4))

axs[0].plot(x, mu)
axs[0].fill_between(x.flatten(), mu.flatten()-K.diag(), mu.flatten()+K.diag(), alpha=0.5)
axs[0].set_xlim([-1, 1])
axs[0].set_ylim([-3, 3])
axs[0].set_title('GP prior')


xtrain = torch.tensor([-0.5, 0.0, 0.75])[:, None]
ytrain = torch.tensor([-1.5, 1.0, 0.5])[:, None]

def posterior_pred(x, xt, yt, gamma=10.0, c=1.0):
    Kxxt = rbf_kernel(x, xt, gamma, c)
    Kxt = rbf_kernel(xt, xt, gamma, c) + torch.eye(xt.shape[0])*s2
    Kinv = torch.linalg.inv(Kxt)
    Kxx = rbf_kernel(x, x, gamma, c)

    mu = Kxxt @ Kinv @ yt
    cov = Kxx - Kxxt @ Kinv @ Kxxt.T
    return mu, cov

post_mu, post_cov = posterior_pred(x, xtrain, ytrain)
axs[1].plot(x, post_mu)
axs[1].fill_between(x.flatten(), post_mu.flatten()-post_cov.diag(), post_mu.flatten()+post_cov.diag(), alpha=0.5)
axs[1].scatter(xtrain, ytrain, color='red', zorder=5)

axs[1].set_xlim([-1, 1])
axs[1].set_ylim([-3, 3])
axs[1].set_title('GP posterior')
```
