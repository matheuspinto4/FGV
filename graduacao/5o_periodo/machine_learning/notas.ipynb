## **3. Métodos de vizinhos mais próximos**

### **3.1 k-NN para classificação**

- Dado o conjunto **$\mathcal{D} = \set{(x_1, y_1), ..., (x_N, y_N)}$**, o rótulo para um vetor qualquer **$x$** pode ser estimado com k-NN seguindo os seguintes passos:
    - 1. compute **$d_n := d(x,x_n)$** para **$n = 1, ..., N$**;
    - 2. ache os índices **$v_1, ..., v_k$** dos $k$ menores elementos de **$d = [d_1, ..., d_N]$** ;
    - 3. retorne **$\hat{y} = h(x) = moda(y_{v_1}, ..., y_{v_k})$**

#### **3.1.2 k-NN para regressão**

- Podemos também utilizar o k-NN para regressão, onde o valor predito **$h(x)$** deverá ser uma média ponderada dos valores dos outros pontos com base na distância.

$$
h(x) = \frac{1}{Z} \sum_{(x^{'}, y^{'}) \subseteq \mathcal{V}_k(x)}{\frac{1}{d(x^{'}, x)}} y^{'} \text{, com } z :=  \sum_{(x^{'}, y^{'}) \in \mathcal{V}_k(x)}{\frac{1}{d(x^{'}, x)}}$$

- O algoritmo não necessita de treinamento, apenas no conjunto de dados (*lazy-learning*).


#### **3.1.2 Qual distância utilizar?** 

##### **Normas $L^p$**

- Também chamada de distância de Minkowski

- **$d_p(x, z) = ||x - z||_p := \left( \sum^{D}_{i=1}{|x_i - z_i|^p} \right)^{\frac{1}{p}}$**
- Não abordam informações sobre a distribuição dos dados

##### **Distância de Mahalanobis**

- **$d_M(x, z) := \sqrt{(x - z)^T \Sigma^{-1} (x - z )}$**
- **$\Sigma = \frac{1}{N - 1} \sum^N_{n = 1}{(x_n - \bar{x})(x_n - \bar{x})}$**


---
**Normalização para média zero e variância um** 

Subtrair a média **$\bar{x}$** de cada vetor e, subsequentemente, multiplicar pela inversa da matriz diagonal **$C$** com entradas **$C_{j,j} = \left( \sum^N_{i = 1}(x_{i,j} - \bar{x})^2 / (N - 1)\right)^{1/2}$** é um procedimento comum em ML.

---

### **3.2 Maldiçao da Dimensionalidade**

A expressão é comumente utilizada para descrever problemas causados pelo aumento exponencial do volume associado em função da dimensionalidade em espaços euclidianos. No caso do K-NN, este exemplo implica na esparcidade dos exemplos de trinamento, onde os k-vizinhos procurados estejam muitos distantes.

Suponha que a distribuição **$\mathbb{P_x}$** sobre os vetores de entrada sejam uniformes sobre uma hiper-bola D-dimensional **$S_D$**, centrada na origem. Queremos classificar o vetor **$0$**, então, o quão grande a bola **$S_D'$**, que contém os k-vizinhos, deve ser?

**$$\mathbb{P}[x_i \in S_D'] = \mathbb{E}_{x_i \sim \mathbb{P_x}}[1_{x_i \in S_D'}] = \left(\frac{\pi^{D/2}}{\Gamma(D/2 + 1)}r^D \right)/\left(\frac{\pi^{D/2}}{\Gamma(D/2 + 1)}1^D\right) = r^D$$**


O número esperado de pontos dentro desta bola é igual a **$Nr^D$**, queremos que est evalor seja igual a **$k$**, então **$r = (k/N)^{1/D}$**. Ou seja, o valor do raio tende a **$1$** a medida que **$D$** cresce.

Isso implica que os vizinhos mais próximos podem estar tão longes quanto a casca esférica. Precisaríamos que o valor de **$N$** crescesse exponencialmente com o valor de **$D$**.

## **4. Modelos lineares para regressão**

### **4.1 Regressão linear**

suponha que as saídas podem ser escritas como uma combinação à fim da respectiva entrada 

**$$y_n = x_n^T \theta + \epsilon_n$$**

Uma maneira de escolher uma estimativa **$\hat{\theta}$** é um que minimize a soma dos erros quadrados em relação às predições feitas pelo modelo linear **$h$**

**$$\hat{\theta}_{LS} = \argmin_{\theta \in \R^{D+1}}\set{l(\theta) := \frac{1}{N} \sum_{n = 1}^{N}(y_n - \theta^Tx_n)^2}$$**

A função de custo também pode ser escrita em forma matricial:

**$$l(\theta) = \frac{1}{N} ||y - X\theta||_2^2$$**

Podemos então obter uma solução ótima achando o gradiente em relação à **$\theta$**

**$$\hat{\theta} = (X^TX)^{-1}X^T y$$**

Caso **$X^TX$** não seja poditiva definida, podemos somar **$\alpha I$** a ela, sendo **$\alpha$** um escalar pequeno.

#### **4.1 Uma perspectiva probabilística**

Podemos obter o mesmo estimador de mínimos quadrados para este modelo:

**$$y_n | x_n ~ N (\theta^Tx_n, \sigma^2) \text{ } \forall n = 1, ..., N$$**

onde nós fazemos o estimador de máximo verssimilhança

### **4.2 Modelo como expansão de base**

Podemos realizar operações não lineares nos modelos de entrada. Transformando o espaço de **$\R^D \rightarrow \R^M$**, em que normalmente **$M$** é maior que **$D$**. Ou seja, o modelo preditivo é dado por

**$$\hat{y} = \theta^T \phi(x) = \theta^T z$$**

Existem várias formas de realizarmos as transformações não lineares

#### **4.2.1 Polinômios**

Podemos utilizar funções de base que são polinômios, ou seja, da forma

**$$\phi(x) = [x^n, ..., 1] $$**
**$$\hat{y} = \theta_1 x^n + ... + \theta_n = \theta^Tz$$**

#### **4.2.2 Funções de base radiais**

De forma simples, consiste em escolher $M$ pontos $c_1, c_2, ..., c_M \in \R^D$, também chamados de centros ou protótipos, e então criar o vetor de regressores $z = \phi(x)$, combinando funções radiais em torno dos centros.

Definição de função radial: Dada uma métrica **$||.||$** no $\R^D$ e $c \in \R^D$, dizemos que $f_c: \R^D \rightarrow \R$ é radial se existe uma função $f:[0, \inf) \rightarrow \R$ tal que
$$f_c(x) = f(||x - c||)$$

Podemos então construir uma transformação $\phi$ que utiliza $M$ funções radiais da forma:

$$\phi(x) = [f_{c_1}(x), ..., f_{c_M}(x)]^T$$

Quando as funções são linearmente independentes, essas funções são chamadas de funções de base radiais (Radial Basis Functions, RBFs).

## **5. Modelos lineares para classificação**

Agora vamos estudar os modelor lineares para o caso em que a variável resposta $y$ é uma classe.

#### **5.1 Regressão Logística**


Se $y$ é uma variável binária (0/1), a escolha mais comum é utilzar uma distribuição Bernoulli com parâmetro **$r = g(x)$**. Ou seja, $y$ assume valor $1$ com probabilidade $r$ e $0$ com probabilidade $1 - r$. Adotando esse modelo observacional para $y | x$, definimos uma função **$g$** linear em **$x$**, mas aplicamos a sigmoide **$\sigma(t) = (1 + e^{-t})^{-1}$** para definir a probabilidade de **$y | x$** como:

**$$p(y|x) = Ber(y | \sigma(\theta^Tx)) \\ = \sigma(\theta^Tx)^y(1 - \sigma(\theta^Tx))^{1 - y}$$**


---
**Propriedades da função sigmoide**


- Estritamente crescente e está contida no intervalo [0,1]
- $\sigma(-t) = 1 - \sigma(t)$
- $\frac{d \sigma(t)}{dt} = \sigma(t)\sigma(-t)$
- $\sigma(t) = \frac{1}{2} + \frac{1}{2}\tanh(\frac{t}{2})$
- $\int\sigma(t)dt = log\sigma(-t) + C$

---

Dado um conjunto de treinamento $\mathcal{D}$ com **N** exemplos de treinamento **$(x_n, y_n)$**, podemos definir a função de máximo verossimilhança como:

**$$\mathcal{L}(\theta) = \Pi_{i = 1}^N \sigma(\theta^Tx_i)^{y_i}(1 - \sigma(\theta^Tx))^{1 - y_i}$$**

Porém, para fins de implementação, nós podemos definir a variável **$y_i' = 2y_i - 1$**. Com isso, conseguimos escrever o gradiente da log verossimilhança para o $i$-ésimo exemplo de treinamento como:

**$$\nabla_{\theta}l_i(\theta) = \nabla_{\theta} \log \sigma(y_i'\theta^Tx) \\ \sigma(-y_i'\theta^Tx)y_i'x_i$$**


---
**Entropia cruzada binária**

O negativo da verossimilhança Bernoulli é referido como a entropia cruzada binária. De forma geral, a entropia entre duas funções de densidade $p$ e $q$ sobre a mesma variável aleatória $z$ e com suportes identicos é dada por:

**$$H(p,q) = \mathbb{E}_{z\sim p}\left[ \log \frac{1}{q(z)}\right]$$**

e dá-se o nome especial BCE para o caso em que $p$ e $q$ são distribuíções Bernoulli.

---

#### **5.2 Regressão Logística Bayesiana**


Um problema com estimativas pontuais, que utilizam os estimadores de máximo verossimilhança, é que elas não refletem a nossa incerteza sobre o valor estimado. \
A estatística Bayesiana propõe uma saída intuitiva para esse empasse. A ideia é modelar os parâmetros **$\theta$** do modelo como uma variável aleatória, usando uma distribuição a priori **$p(\theta)$** e aplicar a regra de Bayes para computar a distribuição de $\theta$ condicionada nas observações $\mathcal{D}$, que chamamos de posteriori:

**$$p(\theta|\mathcal{D}) = \frac{p(\mathcal{D}|\theta)p(\theta)}{\int_{\theta}p(\mathcal{D|\theta})p(\theta)}$$**

Quando não possuimos informação significante sobre os dados, o comum é escolher uma distribuição a priori de alta entropia.

No caso da regressão logística, o comum é adotar uma priori Gaussiana sobre $\theta$ e a verossimilhança anteriormente usada.

**$$\theta \sim \mathcal{N}(\mu, \Sigma) \\ y_n | x_n \sim Ber(\sigma(\theta^Tx_n) \forall n =1, ..., N$$**

Cuja posteriori é dada por:

**$$p(\theta|\mathcal{D}) = \frac{\Pi_{i = 1}^{N} Ber(y_i| \sigma(\theta^Tx_i)) \mathcal{N}(\theta|\mu , \Sigma)}{\int _{x \in \R^{D+1}} \Pi_{i = 1}^{N} Ber(y_i| \sigma(\theta^Tx_i)) \mathcal{N}(\theta|\mu , \Sigma)}$$**


Computar a posteriori acima depende da resolução de um integral que não possui forma fechada. Consequentemente, não há uma forma analítica para a posteriori. É necessário driblar este problema aproximando a função. 


---
**Prevendo a label $y^*$ para um novo input $x^*$.**

Suponha que conseguimos computar a posteriori **$p(\theta|\mathcal{D})$**, podemos obter uma distribuição **$p(y^*|x^*)$** levando em conta a nossa incerteza sobre **$\theta$**:

**$$p(y^*|x^*) = \int_{\theta \in \R^{D+1}} p(y^*|x^*, \theta)p(\theta|\mathcal{D})$$**




---

**Máxima verosimilhança e máximo a posteriori**

Uma das maiores virtudes do paradigma Bayesiano é nos oferecer uma maneira de quantificar a incerteza. Porém nem sempre é viável computar o a posteriori, mesmo que de maneira aproximada. Nestes casos, é comum procurar o ponto que máximiza a posteriori e tomá-lo como estimativa pontual. Este procedimento é chamado de máximo a posteriori (MAP):

**$$\theta_{MAP} = \argmax_{\theta} \set{\log{p(\theta|\mathcal{D})}} \\ = \argmax_{\theta} \set{\log {p(\mathcal{D}|\theta)} + \log p(\theta)}$$**

---

##### **5.2.1 Aproximação de Laplace**

A ideia é construir uma aproximação simples **$q(\theta)$** para a posteriori **$p(\theta|\mathcal{D})$** usando uma expansão de Taylor de segunda ordem em **$\log{p(\theta|\mathcal{D})}$** em torno da moda **$m$** da posteriori:

**$$\log {p(\theta|\mathcal{D})} \approx \log{p(m | \mathcal{D})} + (\theta-m)^T\nabla_{\theta}\log{p(\theta|\mathcal{D})}|_{\theta=m} + \frac{1}{2}(\theta-m)^T\nabla^2_{\theta}\log{p(\theta|\mathcal{D})}|_{\theta=m}(\theta-m) \\ \log{p(\theta|\mathcal{D})} \approx -\frac{1}{2}(\theta-m)^TH(\theta-m) + constante$$**

Agora nós escolhemos uma **$q(\theta)$** que seja proporcional à expressão acima:

**$$\log{q(\theta)} = -\frac{1}{2}(\theta - m)^T H (\theta - m) + constante \Rightarrow q(\theta) \propto e^{-\frac{1}{2} (\theta - m)^T H (\theta- m)}$$**

Logo, temos que **$q(\theta)$** é proporcional a uma normal multivariada com média **$m$** e matriz de covariância igual a inversa de **$H$**:

**$$q(\theta) = \mathcal{N}(\theta|\mu=m, \Sigma = H^{-1})$$**

#### **5.3 Expansão para problemas de multiclasse**

Primeiro, calculamos um vetor de logits **$z$**, desta vez usando uma transformação linear para cada uma das **$L$** classes:

**$$z = \begin{bmatrix} x^T\theta^{(1)} \\ x^T\theta^{(2)} \\ ... \\ x^T\theta^{(L)} \end{bmatrix}$$** 

Agora aplicamos a função softmax para transofrmar **$z$** em um vetor de probabilidades e obter **$r$**, que é dado por:

**$$r = Softmax(z) = \frac{e^z}{\sum^L_{l = 1}e^z_l}$$**

## **6. Inferência variacional**

A ideia é aproximar uma distribuição complexa **$p$**, uma posteriori, com uma distribuição mais simples **$q$**. No entanto, ao invés de fazermos isso com uma expansão de Taylor, nós escolhemos **$q$** dentro de uma família de distribuições **$Q$** de maneira a minimizar uma medida de discrepância para **$p$**. A mais popular destas medidas é a divergência de Kullback-Leibler (KL), definida como:

**$$D_{KL}(q ||p) = \mathbb{E}_{\theta \sim q}\left[\log{\frac{q(\theta)}{p(\theta)}}\right] = \int_{\Theta}{q(\theta) \log{\frac{q(\theta)}{p(\theta)} d\theta}}$$**

### **6.1 Propriedades da divergência KL**

- A divergência KL é não-negativa;
- **$D_{KL}(q||p)$** é zero somente se **$q = p$**;



---

**Divergência KL *foward vs reverse***

É importante notar que, de forma geral, a divergência KL não é simétrica - i.e., **$D_{KL}(p||q) \neq D_{KL}(q||p)$**. É comum dar nome a essas duas divergências:
- **$D_{KL}(q||p)$**: divergência KL **reverse**; 
    - Ela tenta ao final buscar as modas;
- **$D_{KL}(p||q)$**: divergência KL **forward**;
    - Ela tenta ao final ser mais homgênia e se adequar ao todo um pouco;

Como normalmente não temos a distribuição real **$p$**, é mais útil utilizar a versão *reverse*.

### **6.2 Cota Inferior na log-evidência**

No contexto de inferência Bayesiana, $p$ costuma ser a posteriori **$p(\theta|\mathcal{D})$**, que desejamos aproximar.

**$$D_{KL}(q||p(\theta|\mathcal{D})) = \mathbb{E}_{\theta \sim q}[\log q(\theta)] - \mathbb{E}_{\theta \sim q}[\log{p(\theta|\mathcal{D})}] + \log p(\mathcal{D})$$**

Podemos definir **$-L(q) = \mathbb{E}_{\theta \sim q}[\log q(\theta)] - \mathbb{E}_{\theta \sim q}[\log{p(\theta|\mathcal{D})}]$**, logo

**$$D_{KL}(q||p(\theta|\mathcal{D})) = -L(q) + \log p(\mathcal{D}) $$** **$$D_{KL}(q||p(\theta|\mathcal{D})) + L(q) = \log p(\mathcal{D}) = constante$$**

Ou seja, caso queiramos minimizar o valor da distância KL, podemos maximizar o valor da **$L(q)$**, que é conhecida como ELBO.
