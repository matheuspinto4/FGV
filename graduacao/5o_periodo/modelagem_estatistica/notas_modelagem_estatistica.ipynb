#### **6.1 Introdução**

Alguns modelos podem ter este formato:

**$$E(Y_i) = \mu_i = x_i^T \beta\text{; } \text{ } Y_i \sim N(\mu_i, \sigma^2)$$**

onde **$Y_1, ..., Y_n$** são variáveis aleatórias independentes.
Este modelo geralmente é escrito sob a forma

**$$y = X\beta + e$$**

onde **$e \sim N(0, \sigma^2 I)$**. Regressões lineares múltiplas, analises de variância (ANOVA) e análise de covariância (ANCOVA) são todas desta forma e juntas são chamadas de **modelos lineares gerais** (GLM).

#### **6.2 Resultados Básicos**

##### **6.2.1 Estimador de Máximo Verssimilhança**

O estimador de máximo verossimilhança de **$\beta$** é dado por
**$$b = (X^TX)^{-1}X^Ty$$**
como **$E(b) = \beta$**, o estimador é não enviesado

##### **6.2.2 Estimando os mínimos quadrados**

Se **$E(y) = X\beta$** e **$E[(y - X\beta)(y - X\beta)^T] = V$**, onde **$V$** é a matriz de Variância-Covariância, podemos obter o estimador de **$\beta$** sem assumir mais nada sobre a distribuição de **$y$**, basta minimizar

**$$S_w = (y - X\beta)^TV^{-1}(y - X\beta)$$**

onde a solução passa a ser

**$$\widetilde{\beta} = (X^TV^{-1}X)^{-1}X^TV^{-1}y$$**

Em particular, quando os elementos de **$y$** são independentes e tem uma variância comum (homocedatiscidade), 

**$$V = \sigma^2 I$$**
**$$\Rightarrow \widetilde{\beta} = (X^TX)^{-1}X^Ty$$**

##### **6.2.3 Desvio**

Como o vetor dos resíduos é dado por **$e = y - Xb$**, então a soma dos quadrados dos resíduos é dado por **$e^Te$** e o desvio é

**$$D = \frac{1}{\sigma^2} (y - Xb)^T(y - Xb)$$**

##### **6.2.4 Teste de Hipótese**

Antes, é melhor introduzirmos uma nova distribuição, a distribuição F:

**$$F = \frac{\chi^2(d_1) / d_1}{\chi^2(d_2) / d_2}  \sim F(d_1, d_2)$$**

Ela surge desta fração de duas variáveis chi-quadrado independentes.

Considere que existe uma hiótese nula **$H_0$** e uma hipótese mais geral **$H_1$** definidas por

**$$H_0: \beta = \beta_0 \in \R^q$$**
**$$H_1: \beta = \beta_1 \in \R^p$$**

onde **$q < p < N$**. Como a hipótese **$H_1$** é mais geral, o seu erro sempre será menor ou igual ao da hipótese **$H_0$**, já que tenderá a se encaixar melhor aos dados.\
A hipótese **$H_0$** pode ser testada contra **$H_1$** utilizando os desvios **$D_0 \text{ e } D_1$** e a estatística F.

**$$F = \frac{D_0 - D_1}{p - q} / \frac{D_1}{N - p}$$**

Pelo teorema de Cochran, **$D_1 \sim \chi^2{(N - p)}$**. \
O mesmo vale para **$D_0$**, porém devemos considerar duas possibilidades:



- **$H_0$ correta**

    - **$D_0 \sim \chi^2(N - q)$**.
    - **$\rightarrow \nabla D \sim X^2(p - q)$** 
    - **$\rightarrow$** a estatística F seguirá uma distribuição **$\sim F(p - q, N - p)$**

- **$H_0$ incorreta**

    - **$D_0$** pode ter uma distribuição não centralizada **$\chi^2{(N - p, v)}$**. 

    - **$\rightarrow \nabla D = D_0 - D_1$** teria uma distribuição não centralizada **$\chi^2(p - q, v)$** 
    - F teria uma distribuição não centralizada 

Então, os valores de F que são grandes relativos à distribuição **$F(p - q, N - p)$** nos dão evidências de que **$H_0$** é falsa.

##### **6.2.5 Ortogonalidade**

Geralmente inferências feitas sobre um parâmetro depende em quais outras variáveis estão presentes no modelo. Um exceção ocorre quando a matriz **$X$** é ortogonal, onde é possível analisar os parâmetros separadamente.

Na prática, a matriz **$X$** raramente é ortogonal. Então, inferencias sobre qualquer subconjunto de parâmetros **$\beta_j$** depende na ordem a qual os termos são incluídos no modelo. Para se livrar deste problema, muitos programas estatísticos utilizam abordagens de testes que consideram primeiramente que todos os outros termos foram incluídos antes de **$X_j\beta_j$**

##### **6.2.6 Resíduos**

A matriz de variância-covariância do vetor dos residuais **$\hat{e}$** é

**$$E(\hat{e}\hat{e}^T) = E[(y - Xb)^T(y - Xb)] \\
\\
= \sigma^2[I - X(X^TX)^{-1}X^T] = \sigma^2[I - H]$$**

Então os resíduos normalizados é dado por

**$$r_i = \frac{\hat{e}_i}{\hat{\sigma}(1 - h_{ii})^{1/2}}$$**

Estes resíduos devem ser usados para conferir a adequação dos modelos ajustado. 

##### **6.2.7 Outros diagnósticos**

- Um **outlier** é um dado que não é muito bem ajustado pelo modelo. 
- Uma **observação influente** é aquela que tem grande participação nas inferências feitas pelo modelo. 
- Um dado pode acabar não tendo essas duas características ao mesmo tempo

O valor **$h_{ii}$** é chamado de **influência** da $i$-ésima observação. Uma observação com uma alta influência terá um grande impacto no ajuste de um modelo.

Existem estatísticas que combinam os resíduos normalizados e a influência das observações:

**$$DFITS_i = r_i\left(\frac{h_{ii}}{1 - h_{ii}}\right)^{1/2}$$**

e a **distância de Cook**

**$$D_i = \frac{1}{p}\left(\frac{h_{ii}}{1 - h_{ii}}\right)^{1/2}$$**

Grandes valores dessas estatísticas indicam que a observação **$i$** é influente.

Outra abordagem para analisar a influência de uma observação é ajustar um modelo usando e não usando ela, analisando a diferença que isso faz nas estimações de **$b$**. Por exemplo, a estatística **delta-beta** é definida como

**$$\Delta_i\hat{\beta_i} = b_j - b_{j(i)}$$**

Os delta-betas podem ser combinados, formando a distância de Cook:

**$$D_i = \frac{1}{p\hat{\sigma}^2}(b - b_{(i)})^TX^TX(b - b_{(i)})$$**

Quando uma observação influenciável ou outlier é detectado, o primeiro passo é determinar se ela pode ser um erro de medição, erro de transcrição ou algum outro tipo de erro. Ela deve ser removida do conjunto de dados somente se houver uma razão substancial para tal.

### **6. Por trás da modelagem de regressões**

#### **6.1 Modelos de regressão**

O modelo linear mais simples é aquele com apenas um preditor:

**$$\text{Modelo linear simples: } y = a + bx + erro$$**

#### **6.3 Interpretando coeficientes como comparações, não efeitos**

Nós precisamos ter cuidado ao interpretar os coeficientes de um modelo ajustado de regressão, não podemos passar do ponto ótimo de interpretação e exagerar nas análises.\
A forma mais segura de interpretação é a comparação, onde devemos comparar os pontos dentro dos dados com eles mesmos. \
A regressão é uma ferramenta matemática para realizar previsões. Coeficientes de uma regressão podem às vezes ser interpretados como efeitos, mas sempre podem ser interpretados como comparações médias.



#### **6.5 O paradoxo da regressão à média**

Os pontos de previsão regressam em direção à média - este é o coeficiente angular menor que 1 - e isso reduz a variação. Ao mesmo tempo, o erro do modelo - a imperfeição da previsão - adiciona variação, somente o suficiente para manter a variancia total constante de uma geração para a outra de pontos. \
Regressão para a média sempre vai surgir sempre que as previsões forem imperfeitas em um ambiente estável. A imperfeição da previsão induz uma variação.

O erro humano é tentar achar uma explicação para uma regressão à média que está ocorrendo puramente por conta da estatística e matemática.

### **10. Regressões lineares com múltiplos preditores**

Agora o nosso modelo se torna mais complexo, onde mais de uma variável é incluída na previsão:

**$$y = \beta_0 + \beta_1x_1 + \beta_2x_2 + ... + erro $$**

#### **10.1 Adicionando previsores a um modelo**

Modelos com mais de um previsor são mais difíceis de serem interpretados, pois a interpretação de qualquer coeficiente é, em partes, contingente nas outras variáveis do modelo. O coeficiente **$\beta_k$** é a média ou esperança da diferença na saída **$y$**, comparando dois pontos que se diferenciam somente na variável correpondente **$x_k$** enquanto iguais em todos os outros valores de variáveis.
