### **Busca**

- **Formulação de Problemas:**
    - **Espaço de Estados:** Representa os estados possíveis no mundo
    - **Estado Inicial** Escrito na formatação do Espaço de Estados
    - **Ações:** Ação($s$) = possíveis ações do estado $s$
    - **Teste de objetivo:** Teste para ver se a resposta final já foi alcançada
    -  **Modelo de Transição:** Descreve o que o ambiente faz quando uma ação $a$ é executada no estado $s$ Resultado($s$, $a$)
    -  **Custo de Caminho**: **$c(x, a, y)$**

- **Função Heurística:**
    - Uma função heurística é admissível quando ela nunca superestima o valor do custo real de uma ação.
    - Para saber se uma heurística é admissível, basta dar superpoderes ao agente e imaginar um cenário em que o custo dele seja sempre menor do que o custo real.

### **Lógica**

- **Acarretamento Lógico / Consequência Lógica (⊨)**: Relação onde a verdade das premissas garante a verdade da conclusão.
    - Para ser válido, não pode existir nenhum cenário (modelo) onde a premissa seja Verdadeira e a conclusão seja Falsa.
    - Pode ser demonstrado construindo uma tabela-verdade ou buscando um contraexemplo.
    



- **Satisfatibilidade e Tautologia:** Propriedades que avaliam o resultado de uma sentença.
    - Satisfatível: Existe pelo menos um modelo (combinação de V/F) em que a sentença resulta em Verdadeiro.
    - Tautologia: A sentença resulta em Verdadeiro em absolutamente todos os modelos possíveis (ex: A∨¬A).



- **Modelos Lógicos:** Atribuições de valores lógicos (V ou F) para todo o vocabulário do problema.
    - O número total de modelos em um universo é sempre 2n, onde n é a quantidade de proposições (letras) do vocabulário.
    - Para calcular quantos modelos satisfazem uma sentença, fixa-se o valor das variáveis restritas pela sentença e multiplica-se pelas opções das variáveis "livres" do vocabulário.



- **Tradução de Linguagem Natural:** Conversão de texto para lógica proposicional.
    - "se P então Q" vira P→Q.
    - "P se Q" vira Q→P.
    - "P se e somente se Q" (ou "P se Q, mas de outra forma não P") vira o bicondicional P↔Q.



- **Equivalências Lógicas:** Regras obrigatórias para manipular e simplificar expressões.
    - Implicação: A→B≡¬A∨B
    - Bicondicional: A↔B≡(A→B)∧(B→A)
    - Leis de De Morgan: ¬(A∧B)≡¬A∨¬B e ¬(A∨B)≡¬A∧¬B



- **Regras de Inferência Clássicas:** Mecanismos de dedução passo a passo a partir de verdades conhecidas.
    - Modus Ponens: De A→B e A, conclui-se B.
    - Modus Tollens: De A→B e ¬B, conclui-se ¬A.
    - Silogismo Disjuntivo: De A∨B e ¬A, conclui-se B.


- **Prova por Contradição (Redução ao Absurdo):** Método de demonstração lógica.
    - Assume-se o inverso (a negação) do que se deseja provar.
    - Aplica-se as regras de inferência em cadeia.
    - O objetivo é encontrar uma impossibilidade lógica (ex: provar A e ¬A ao mesmo tempo), confirmando que a suposição inicial estava errada.



- **Cláusulas de Horn:** Um tipo restrito de cláusula muito usado em IA.
    - É sempre uma disjunção (termos separados apenas por ∨).
    - Possui, no máximo, um literal positivo (sem o símbolo ¬). Exemplo: ¬A∨¬B∨C.

- **Método por Resolução:**
    - Nege a hipótese
    - Adicione ela a sua base de conhecimento
    - Passe sua base de conhecimento para a FNC
    - Inicie os cancelamentos das cláusulas
    - Pare se achar uma cláusula vazia, isto prova a hipótese. Caso não chegue em uma contradição, a hipótese é inválida.


- **FNC:** Forma Normal Conjuntiva
    - Só podem existir três conectores lógicos.
    - **$\lor \text{ e } \neg$** dentro dos parênteses
    - **$\land$** fora dos parênteses

**$$f(n) = g(n) + h(n)$$**

onde **$g(n)$** é o custo do caminho desde o estado inicial até o nó **$n$**, e **$h(n)$** é o custo estimado do caminho mais curto.

**Heurística Admissível**

Uma heurística é admissível quando nunca superestima o custo de atingir a meta.

**Consistência**

Uma heurística é consistênte quando, para cada nó $n$ e para cada sucessor $n'$ de $n$ gerado por uma ação $a$, tivermos

**$$h(n) \le c(n, a, n')+ h(n')$$**

### **7.4 Lógica Proposicional: uma lógica muito simples**

As **sentenças atômicas** consistem num único **símbolo proposicional**. Cada um desses símbolos representa uma proposição que pode ser verdadeira ou falsa. As **sentenças complexas** são formadas a partir da combinação destes símbolos.

Existem cinco conectivos lógicos de uso comum:

- **Negação Literal:** **$\neg W$** é chamada de negação de $W$. Um **literal** é uma sentença atômica (um **literal positivo**) ou uma sentença negada (um **literal negativo**).
- **Conjunção: $W \land P$** é uma **conjunção**, suas partes são os **elementos da conjunção**. 
- **Disjunção: $W\lor P$** é uma **disjunção**, suas partes são os disjuntos.
- **Implicação: $W \implies P$** (implica) é uma **implicação**. A premissa é **$W$** enquando a conclusão é **$P$**.
- **Bicondicional: $W\iff P$** (se e somente se) é uma **bicondicional**.

**Precedência de operadores**: **$\neg, \land, \lor, \implies, \iff$**

#### **7.4.2 Semântica**

Para sentenças complexas temos cinco regras:

- **$\neg P$** é verdadeiro, sse $P$ for falso em $m$
- **$P \land Q$** é verdadeiro, sse $P$ e $Q$ forem verdadeiros em $m$
- **$P \lor Q$** é verdadeiro, sse $P$ ou $Q$ for verdadeiro em $m$
- **$P \Rightarrow Q$** é verdadeiro , a menos que $P$ seja verdadeiro e $Q$ seja falso em $m$
- **$P \iff Q$** é verdadeiro sse $P$ e $Q$ forem ambos verdadeiros ou ambos falsos em $m$
