# **Exercício 1 Produto Interno e Gram-Schmidt em $P_2$**



O espaço $P_2$ consiste nos polinômios $p(x) = ax^2 + bx + c$. A operação definida é:
$$\langle p, q \rangle = \int_{-1}^1 p(x)q(x)dx$$



##### **(a) Mostrando que é um produto interno: Para que essa operação seja um produto interno, ela deve satisfazer quatro axiomas para quaisquer polinômios $p, q, r \in P_2$ e escalar $\alpha \in \mathbb{R}$:**


- 1 Simetria: $$\langle p, q \rangle = \int_{-1}^1 p(x)q(x)dx = \int_{-1}^1 q(x)p(x)dx = \langle q, p \rangle$$
- 2 Linearidade na primeira variável: $$\langle \alpha p + r, q \rangle = \int_{-1}^1 (\alpha p(x) + r(x))q(x)dx = \alpha \int_{-1}^1 p(x)q(x)dx + \int_{-1}^1 r(x)q(x)dx = \alpha \langle p, q \rangle + \langle r, q \rangle$$
- 3 Positividade: $$\langle p, p \rangle = \int_{-1}^1 (p(x))^2 dx \ge 0$$(A integral de uma função não negativa é sempre maior ou igual a zero).
- 4 Definitividade: Se $\langle p, p \rangle = 0$, então $\int_{-1}^1 (p(x))^2 dx = 0$. Como $(p(x))^2$ é uma função contínua e não negativa no intervalo $[-1, 1]$, a única forma da área sob a curva ser zero é se a função for identicamente nula nesse intervalo. Logo, $(p(x))^2 = 0 \implies p(x) = 0$.

Como os quatro axiomas são satisfeitos, a operação é um produto interno autêntico

##### **(b) Base ortonormal via Gram-Schmidt**

Vamos partir da base canônica de $P_2$, que é $\beta = \{1, x, x^2\}$. Chamaremos esses vetores de $u_1 = 1$, $u_2 = x$ e $u_3 = x^2$. O processo de Gram-Schmidt gerará vetores ortogonais $v_i$, que depois normalizaremos para encontrar $q_i$.

- Passo 1: O primeiro vetor $$v_1 = u_1 = 1$$
Calculando sua norma:$$\|v_1\|^2 = \langle 1, 1 \rangle = \int_{-1}^1 1 dx = 2 \implies \|v_1\| = \sqrt{2}$$
O primeiro vetor ortonormal é:$$q_1 = \frac{v_1}{\|v_1\|} = \frac{1}{\sqrt{2}}$$

- Passo 2: O segundo vetor $$v_2 = u_2 - \text{proj}_{v_1}u_2 = u_2 - \frac{\langle u_2, v_1 \rangle}{\|v_1\|^2}v_1$$ 
O produto interno é $\langle x, 1 \rangle = \int_{-1}^1 x dx = 0$ (função ímpar em intervalo simétrico).
Logo, $v_2 = u_2 - 0 = x$.
Calculando a norma:$$\|v_2\|^2 = \langle x, x \rangle = \int_{-1}^1 x^2 dx = \left[ \frac{x^3}{3} \right]_{-1}^1 = \frac{2}{3} \implies \|v_2\| = \sqrt{\frac{2}{3}}$$
O segundo vetor ortonormal é:$$q_2 = \frac{v_2}{\|v_2\|} = \frac{x}{\sqrt{2/3}} = \sqrt{\frac{3}{2}}x$$

- Passo 3: O terceiro vetor $$v_3 = u_3 - \text{proj}_{v_1}u_3 - \text{proj}_{v_2}u_3$$
$$v_3 = x^2 - \frac{\langle x^2, 1 \rangle}{\|v_1\|^2}v_1 - \frac{\langle x^2, x \rangle}{\|v_2\|^2}v_2$$
Calculando os produtos internos necessários:$\langle x^2, 1 \rangle = \int_{-1}^1 x^2 dx = \frac{2}{3}$
$\langle x^2, x \rangle = \int_{-1}^1 x^3 dx = 0$ (função ímpar)Substituindo na fórmula de $v_3$:
$$v_3 = x^2 - \frac{2/3}{2}(1) - 0 = x^2 - \frac{1}{3}$$
Calculando a norma:$$\|v_3\|^2 = \int_{-1}^1 \left(x^2 - \frac{1}{3}\right)^2 dx = \int_{-1}^1 \left(x^4 - \frac{2}{3}x^2 + \frac{1}{9}\right) dx = \left[ \frac{x^5}{5} - \frac{2x^3}{9} + \frac{x}{9} \right]_{-1}^1$$
$$\|v_3\|^2 = 2 \left( \frac{1}{5} - \frac{2}{9} + \frac{1}{9} \right) = 2 \left( \frac{1}{5} - \frac{1}{9} \right) = 2 \left( \frac{4}{45} \right) = \frac{8}{45}$$
$$\|v_3\| = \sqrt{\frac{8}{45}} = \frac{2\sqrt{2}}{3\sqrt{5}}$$
O terceiro vetor ortonormal é:$$q_3 = \frac{v_3}{\|v_3\|} = \frac{3\sqrt{5}}{2\sqrt{2}}\left(x^2 - \frac{1}{3}\right) = \frac{\sqrt{10}}{4}(3x^2 - 1)$$
A base ortonormal é $\{ \frac{1}{\sqrt{2}}, \sqrt{\frac{3}{2}}x, \frac{\sqrt{10}}{4}(3x^2 - 1) \}$.

```python
import numpy as np
import pandas as pd
from scipy.linalg import null_space

df = pd.read_csv('MatrizIncidencia.csv', header=None, sep=';')

M = df.values.astype(float)

# (a) Encontra a base para o núcleo
base_nucleo = null_space(M)

print("=== LETRA (A) ===")
print("Dimensão do Núcleo (Número de colunas na base):", base_nucleo.shape[1])
print("\nVetor(es) da Base do Núcleo:\n", np.round(base_nucleo, 4))

# (b) Identificação das Componentes Conexas
print("\n=== LETRA (B) ===")
num_componentes = base_nucleo.shape[1]
print(f"A rede possui {num_componentes} componente(s) conexa(s).")

# Extraindo os vértices de cada componente com base nos valores não-nulos
for i in range(num_componentes):
    vetor_base = np.abs(base_nucleo[:, i])
    vertices = np.where(vetor_base > 1e-10)[0] + 1
    print(f"Componente {i+1} contém os vértices: {list(vertices)}")
```
